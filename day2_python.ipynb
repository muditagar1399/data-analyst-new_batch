```python
# prime numbers are the numbers having only factor of 1 other than itself for ex : 2, 3, 5
```

```python
num=int(input("enter a number : "))
if num<=1:
    print("not prime")
else:
    for i in range(2,num):
        if num%i==0:
            print("not prime")
            break
    else:
        print("prime")
```
