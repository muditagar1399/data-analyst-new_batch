    # IQR(INER QUARTILE RANGE ) it measures the spread of data middle of data 50% it is commanly used for finding outliers
    # IQR= Q3-Q1
    #Q3 = FIRST QUARTILE (25% data 25th percentile - 25% of values are below it )
    # Q2= MEDIAN = 50 TH PERCENTILE 
    # Q3= 75TH percentile = 75 percent values are below it 

# how to check outliers using IQR

```python
# FIRST IQR (Q3-Q1)
```

# Lower bound = Q1-1.5*1OR
# UPPER BOUND = Q3+1.5*IQR 

# ANY VALUE BELOW LOWER BOUND - OULIER 
# ANY VALUE ABOVE UPPER BOUND - OUTLIER 

```python
# EXAMPLE
# IQR= 300-200=100
#LOWER BOUND = 200-(1.5*100)
#UPPER BOUND = 300 +(1.5*100)
```

```python
# data
```

```python
Q1= data['Sales'].quantile(0.25)
Q3= data['Sales'].quantile(0.75)
IQR=Q3-Q1
print("Q1: ", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
print("Lower BOUND: ", lower_bound)
print("upper bound :", upper_bound)
```

```python
# | is the or operator 
```

```python
outliers=data[
    (data['Sales']<lower_bound)|
    (data['Sales']>upper_bound)
    ]
print(outliers)
```
