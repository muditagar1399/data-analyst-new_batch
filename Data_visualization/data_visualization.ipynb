```python
deposit_counts=data['deposit'].value_counts()
plt.figure(figsize=(12,5))
bars=plt.bar(
    deposit_counts.index,
    deposit_counts.values,
    color=['purple','orange'],
    edgecolor='blue',
    width=0.6
)
plt.title('Deposit Subscription Distribution', 
          fontsize=16,fontweight='bold')
plt.xlabel('Deposit Subscription', fontsize=12)
plt.ylabel('Number of Customers ' , fontsize=12)

plt.bar_label(bars,padding=5,fontsize=11)
plt.grid(axis='y',linestyle='--',alpha=0.3)
plt.tight_layout()
plt.show()
     
```

```python
# # function and meaning 
# Meaning of every function
# Function	Definition
# value_counts()	Counts the occurrences of each unique value in a column.
# plt.figure()	Creates a new figure or canvas for the chart.
# plt.pie()	Creates a pie chart to represent proportions or percentages of a whole.
# plt.title()	Adds a title to the chart.
# plt.tight_layout()	Automatically adjusts spacing to prevent chart elements from overlapping or being cut off.
# plt.show()	Displays the completed chart.
# Meaning of every parameter and attribute
# Parameter / Attribute	Definition
# figsize	Sets the figure's width and height in inches.
# values	Provides the numerical values used to determine the size of each pie slice.
# labels	Displays category names beside the pie slices.
# autopct	Displays percentages on the pie slices.
# startangle	Sets the starting angle of the first pie slice.
# colors	Specifies the colors of the pie slices.
# explode	Separates selected slices from the center of the pie chart.
# wedgeprops	Customizes the appearance of pie slices, including their borders.
# edgecolor	Specifies the border color of the pie slices.
# linewidth	Controls the thickness of the slice borders.
# textprops	Customizes the appearance of text displayed on the pie chart.
# fontsize	Sets the size of text.
# fontweight	Sets the thickness of text, such as bold or normal.
# autopct='%1.1f%%'	Formats percentages to display one digit after the decimal point.
# startangle=90	Starts drawing the pie chart from 90 degrees.
# explode=(0.04, 0)	Moves the first slice slightly outward while keeping the second slice in its original position.

```

```python
deposit_counts=data['deposit'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(
    deposit_counts.values,
    labels=deposit_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['yellow','orange'],
    explode=(0.04,0),
    wedgeprops={'edgecolor':'blue','linewidth':2},
    textprops={'fontsize': 12}
)
plt.title('Share of the customers by deposit status', fontsize=12,fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# ###function and meaning 

# #Function	Definition
# groupby()	Groups rows that have the same value in one or more columns.
# count()	Counts the non-null values in a column or group.
# sort_values()	Sorts values in ascending or descending order.
# plt.figure()	Creates a new figure or canvas for a visualization.
# plt.pie()	Creates a pie chart to represent proportions of a whole.
# plt.title()	Adds a title to the chart.
# plt.tight_layout()	Automatically adjusts spacing to reduce overlapping or clipped chart elements.
# plt.show()	Displays the completed chart.
# Meaning of every parameter and attribute
# Parameter / Attribute	Definition
# balance.values	Returns the numerical counts for each balance category.
# balance.index	Returns the balance category names.
# ascending=False	Sorts values in descending order, from largest to smallest.
# figsize	Specifies the figure's width and height in inches.
# autopct	Displays percentages on the pie slices.
# startangle	Specifies the starting angle of the pie chart.
# colors	Specifies the colors used for the pie slices.
# [:len(balance)]	Selects only as many colors as there are balance categories.
# explode	Separates pie slices from the center of the chart.
# [0.04] * len(balance)	Creates a list that moves every slice outward by 0.04.
# shadow	Controls whether a shadow appears beneath the pie chart.
# wedgeprops	Customizes the appearance of the pie slices.
# edgecolor	Specifies the border color of the slices.
# linewidth	Controls the thickness of the slice borders.
# textprops	Customizes the appearance of text on the chart.
# fontsize	Specifies the font size.
# fontweight	Specifies the thickness of the font, such as bold.
# pad	Controls the spacing between the chart title and the chart area.
# Meaning of the Pandas expressions
# Expression	Definition
# data['deposit'] == 'yes'	Checks which rows have a deposit status of yes.
# data[...]	Filters the DataFrame using a condition.
# data['balance_categroy']	Selects the balance category column.
# data['deposit']	Selects the deposit column.
# len(balance)	Returns the number of balance categories in the resulting Series
```

```python
balance=data[data['deposit']=='yes'].groupby('balance_categroy')['deposit'].count().sort_values(ascending=False)
plt.figure(figsize=(7,7))
plt.pie(
    balance.values,
    labels=balance.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['yellow','orange','pink','red'][:len(balance)],
    explode=[0.04]*len(balance),
    shadow=False,
    wedgeprops={'edgecolor':'blue','linewidth':2},
    textprops={'fontsize': 12}
)
plt.title('Depsoit by balace category ', fontsize=12,fontweight='bold',pad=18)
plt.tight_layout()
plt.show()
```
