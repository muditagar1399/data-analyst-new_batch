```python
# write a program to find factorial of a number 
```

```python
#5!= 5*4*3*2*1=120
```

```python
# factorial meand multiplication a positive integer by all the postiive integers smaller than it it represented by !
#5!= 5*4*3*2*1=120
#120!= 120*119*118...........*2*1
# factorial formula = n!=n*(n-1)*(n-2)*...................*2*1
```

```python
n=int(input("enter a number :"))
fact=1
for i in range(1,n+1):
    fact=fact*i
print("factorial= ", fact)
```
