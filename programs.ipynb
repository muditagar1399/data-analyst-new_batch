# shopping discount : write a program in python to input the total shopping
# amount if the amount is above rs 5000 give a 10% discount display the final bill

```python
amount=float(input("enter the bill amount in rupees :  "))
if amount>5000:
    discount=amount*10/100
else:
    discount=0
final_bill=amount-discount
print("shopping amount : ", amount)
print("discount : ", discount)
print("final_bill: ", final_bill)
```

```python
#1. student grade calckuator

# input marks for 5 subjects
# calculate the total percentage
# display the grade based on percentage
```

```python
marks1=float(input("enter the marks of english : "))
marks2=float(input("enter the marks of hindi : "))
marks3=float(input("enter the marks of science : "))
marks4=float(input("enter the marks of Physics : "))
marks5=float(input("enter the marks of Mathematics : "))
total=marks1+marks2+marks3+marks4+marks5
percentage=(total/500)*100
print("total marks: ", total)
print("percentage: ", percentage, "%")

if percentage>=90:
    print("grade is A")
elif percentage>=80:
    print("grade is B")
elif percentage>=75 and percentage>=70:
    print("grade is B+")
elif percentage>=60:
    print("grade is c")
elif percentage>=50:
    print("Grade is D")
elif percentage>=40:
    print("Grade is E")
else:
    print("Fail")

```

```python
#movie ticket booking : ask the customer age and number of tickets , children , adults and seniors have different ticket prices calculate the total cost
```

```python
age=int(input("enter the age: "))
tickets=int(input("enter number of tickets you have: "))
if age<13:
    price=100
elif age>=60:
    price=120
else:
    price=200
total=price*tickets
print("ticket price : ", price)
print("number of tickets: ", tickets)
print("total cost: ", total)
```

```python
# suppose there are mutliple tickets but each person is of different age 
```

```python
tickets=int(input("enter number of tickets: "))
total=0
for i in range(tickets):
    age=int(input("enter the age of a person " + str(i+1) + ": "))

    if age<13:
        price=100
    elif age>=60:
        price=120
    else:
        price=200

    total=total+price
print("total cost : " , total)
```
